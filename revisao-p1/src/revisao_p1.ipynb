### DESCRITIVA · QUESTÃO 2 DE 32 (Princípio GIGO)
**Enunciado:** Explique o princípio GIGO e por que ele se aplica a esta base.

**Resposta:**
O princípio **GIGO** (*Garbage In, Garbage Out* / "Lixo entra, lixo sai") afirma que se dados brutos de baixa qualidade, incorretos, incompletos ou inconsistentes forem inseridos em um modelo estatístico ou algoritmo de análise, os resultados e conclusões produzidos serão igualmente incorretos e não confiáveis.

Ele se aplica diretamente a esta base porque, se calculássemos métricas (como tarifa média, distância percorrida ou tempo médio de viagem) diretamente do Registro Bruto sem limpeza prévia, as métricas seriam severamente distorcidas: por exemplo, a distância da corrida 8 (3800 km) inflaria desproporcionalmente a média de distância, a duração negativa da corrida 10 geraria tempos médios irreais, e a avaliação 6 distorceria a nota média dos motoristas.

**Enunciado:** Classifique quanto ao tipo de variável: `id_corrida`, `avaliacao`, `tarifa`, `categoria` e `data_hora`. Justifique `avaliacao`.

**Resposta:**
* `id_corrida`: **Qualitativa Nominal** (é um identificador único/código, os números não expressam quantidade real nem ordem).

* `categoria`: **Qualitativa Nominal** (categorias discretas de serviço sem ordem inerente, ex: Comum, Executivo, Moto).
* `data_hora`: **Quantitativa / Temporal (Intervalar ou Contínua)** (representa um instante específico no tempo contínuo).

### CÓDIGO PYTHON · QUESTÃO 8 DE 32 (Duplicatas e datas)
**Enunciado:** Escreva o código para remover a linha duplicada e converter `data_hora` (que vem em 3 formatos) para datetime.

```python
# Questão 8: Remover duplicatas e converter data_hora
# 1. Remover duplicatas
df = df.drop_duplicates().reset_index(drop=True)

# 2. Tratar e converter data_hora
def parse_mixed_dates(val):
    if pd.isna(val):
        return pd.NaT
    val_str = str(val).replace('h', ':').replace('T', ' ')
    return pd.to_datetime(val_str, dayfirst=True, errors='coerce')

df['data_hora'] = df['data_hora'].apply(parse_mixed_dates)
df[['id_corrida', 'data_hora']].head()
```

### CÓDIGO PYTHON · QUESTÃO 9 DE 32 (Número como texto)
**Enunciado:** Escreva o código para converter `tarifa` (ex.: "R$ 14,00", "19,00", vazio) para `float`.

```python
# Questão 9: Converter tarifa para float
df['tarifa'] = (
    df['tarifa']
    .astype(str)
    .str.replace('R$', '', regex=False)
    .str.replace(',', '.', regex=False)
    .str.strip()
)
df['tarifa'] = pd.to_numeric(df['tarifa'], errors='coerce')
df[['id_corrida', 'tarifa']].head()
```

### CÓDIGO PYTHON · QUESTÃO 10 DE 32 (Padronização de categorias)
**Enunciado:** Escreva o código para padronizar `cidade`, `categoria`, `forma_pgto` e `status` (ex.: "SQ" / "Saquarema-RJ" -> "Saquarema"; "cartao" -> "Cartão"; "concluida" -> "Concluída").

```python
# Questão 10: Padronização de strings e mapeamento
# Padronizar motorista
df['motorista'] = df['motorista'].astype(str).str.strip().str.title()

# Mapeamentos de padronização
mapa_cidade = {
    'saquarema': 'Saquarema',
    'saquarema-rj': 'Saquarema',
    'sq': 'Saquarema',
    'araruama': 'Araruama',
    'bacaxá': 'Bacaxá',
    'bacaxa': 'Bacaxá'
}

mapa_categoria = {
    'comum': 'Comum',
    'executivo': 'Executivo',
    'exec': 'Executivo',
    'moto': 'Moto'
}

mapa_forma_pgto = {
    'pix': 'Pix',
    'cartao': 'Cartão',
    'cartão': 'Cartão',
    'dinheiro': 'Dinheiro'
}

mapa_status = {
    'concluída': 'Concluída',
    'concluida': 'Concluída',
    'cancelada': 'Cancelada',
    'em andamento': 'Em andamento'
}

df['cidade'] = df['cidade'].astype(str).str.strip().str.lower().map(mapa_cidade).fillna(df['cidade'])
df['categoria'] = df['categoria'].astype(str).str.strip().str.lower().map(mapa_categoria).fillna(df['categoria'])
df['forma_pgto'] = df['forma_pgto'].astype(str).str.strip().str.lower().map(mapa_forma_pgto).fillna(df['forma_pgto'])
df['status'] = df['status'].astype(str).str.strip().str.lower().map(mapa_status).fillna(df['status'])

df[['cidade', 'categoria', 'forma_pgto', 'status']].head()
```

### CÓDIGO PYTHON · QUESTÃO 11 DE 32 (Valores impossíveis)
**Enunciado:** Escreva o código para: distância > 100 (está em metros) -> dividir por 1000; distância e duração <= 0 -> NaN; avaliação fora de [1, 5] -> NaN.

```python
# Questão 11: Tratamento de regras numéricas e impossíveis
# Distância > 100 em metros -> dividir por 1000
df.loc[df['distancia_km'] > 100, 'distancia_km'] = df.loc[df['distancia_km'] > 100, 'distancia_km'] / 1000.0

# Distância e Duração <= 0 -> NaN
df.loc[df['distancia_km'] <= 0, 'distancia_km'] = np.nan
df.loc[df['duracao_min'] <= 0, 'duracao_min'] = np.nan

# Avaliação fora de [1, 5] -> NaN
df.loc[~df['avaliacao'].between(1, 5), 'avaliacao'] = np.nan

df[['id_corrida', 'distancia_km', 'duracao_min', 'avaliacao']]
```

### CÁLCULO · QUESTÕES 16 A 21 (Análise das Tarifas)
**Valores de tarifa considerados (14 corridas concluídas limpas do Registro Bruto):**
9, 10, 14, 15, 15.5, 16, 19, 20, 22, 25, 28, 40, 55, 190.

```python
# Cálculo exato das questões 16 a 21
tarifas = np.array([9, 10, 14, 15, 15.5, 16, 19, 20, 22, 25, 28, 40, 55, 190])

media = np.mean(tarifas)
mediana = np.median(tarifas)
dp_amostral = np.std(tarifas, ddof=1)

q1 = np.percentile(tarifas, 25) # interpolação linear padrão pandas/numpy
q3 = np.percentile(tarifas, 75)
iqr = q3 - q1

limite_inferior = q1 - 1.5 * iqr
limite_superior = q3 + 1.5 * iqr

outliers = tarifas[(tarifas < limite_inferior) | (tarifas > limite_superior)]
tarifas_sem_outliers = tarifas[(tarifas >= limite_inferior) & (tarifas <= limite_superior)]
media_sem_outliers = np.mean(tarifas_sem_outliers)

print(f"Q16 - Média: {media:.2f}")
print(f"Q17 - Mediana: {mediana}")
print(f"Q18 - Desvio-padrão amostral: {dp_amostral:.2f} (Arredondado: {round(dp_amostral)})")
print(f"Q19 - Q1: {q1}, Q3: {q3}, IQR: {iqr}")
print(f"Q20 - Limite sup: {limite_superior}, Outliers encontrados: {outliers} (Quantidade: {len(outliers)})")
print(f"Q21 - Média sem outliers ({len(tarifas_sem_outliers)} tarifas): {media_sem_outliers:.2f}")
```

### CÁLCULO · QUESTÃO 31 DE 32 (Teorema de Bayes)
**Enunciado:** Sabe-se que $P(\text{cancelar} \mid \text{chuva}) = 0,40$, $P(\text{cancelar} \mid \text{sem chuva}) = 0,10$ e $P(\text{chuva}) = 0,30$. Se uma corrida foi **cancelada**, qual é $P(\text{era dia de chuva})$? (2-3 casas)

**Resposta:**
Pelo Teorema de Bayes:
$$P(\text{Chuva} \mid \text{Cancelada}) = \frac{P(\text{Cancelada} \mid \text{Chuva}) \cdot P(\text{Chuva})}{P(\text{Cancelada})}$$

Onde:
- $P(\text{Sem Chuva}) = 1 - 0,30 = 0,70$
- $P(\text{Cancelada}) = (0,40 \times 0,30) + (0,10 \times 0,70) = 0,12 + 0,07 = 0,19$
- $P(\text{Chuva} \mid \text{Cancelada}) = \frac{0,12}{0,19} \approx 0,631579$

```python
p_c_dado_ch = 0.40
p_c_dado_sch = 0.10
p_ch = 0.30
p_sch = 0.70

p_c = (p_c_dado_ch * p_ch) + (p_c_dado_sch * p_sch)
p_ch_dado_c = (p_c_dado_ch * p_ch) / p_c

print(f"Q31 - P(Chuva | Cancelada): {p_ch_dado_c:.3f} (ou {p_ch_dado_c:.2f})")
```

**Resultado Questão 31:** **0.632** (ou **0,63** / **63,2%**).

### DESCRITIVA · QUESTÃO 32 DE 32 (Raciocínio Poisson)
**Enunciado:** Na questão da Poisson, por que P(>= 1) foi calculated como $1 - e^{-\lambda}$ e por que $\lambda$ passou a ser 1,25?

**Resposta:**
1. **Por que $P(X \ge 1) = 1 - e^{-\lambda}$:**
   A probabilidade de ocorrer "pelo menos 1 evento" é o complemento exato de ocorrer "nenhum evento" ($X = 0$). Pela fórmula da Poisson: $P(X = 0) = \frac{\lambda^0 e^{-\lambda}}{0!} = e^{-\lambda}$. Portanto, calcular $1 - P(X = 0) = 1 - e^{-\lambda}$ é muito mais simples e direto do que somar as probabilidades infinitas de $X=1, 2, 3, \dots$.

2. **Por que $\lambda$ passou a ser 1,25:**
   O parâmetro $\lambda$ da distribuição de Poisson representa a **taxa média dentro do intervalo de tempo considerado**. Como a taxa dada era de 5 solicitações por hora (60 minutos) e a pergunta se refere a um intervalo de 15 minutos (que corresponde a $\frac{15}{60} = 0,25$ de uma hora), ajustamos proporcionalmente a taxa média:
   $$\lambda_{15\text{min}} = 5 \times \frac{15}{60} = 5 \times 0,25 = 1,25 \text{ solicitações / 15 min}$$
